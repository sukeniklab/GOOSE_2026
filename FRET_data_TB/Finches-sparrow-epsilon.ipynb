```python
import numpy as np
import pandas as pd
from finches import Mpipi_frontend, CALVADOS_frontend
from sparrow import Protein
```

```python
mf = Mpipi_frontend()
cf = CALVADOS_frontend()
```

```python
seqs = pd.read_csv('sequences.csv')
```

```python
# Create a copy of the original dataframe to start with
epsilons = seqs.copy()

# Initialize lists to store the epsilon values
mpipi_epsilon_values = []
calvados_epsilon_values = []

# Get the column names (first column is identifier, second is sequence)
columns = seqs.columns.tolist()
identifier_col = columns[0]
sequence_col = columns[1]

# Iterate through each sequence and calculate epsilon values
for index, row in seqs.iterrows():
    sequence = row[sequence_col]
    identifier = row[identifier_col]
    
    try:
        # Calculate Mpipi epsilon (self-interaction)
        mpipi_eps = mf.epsilon(sequence, sequence)
        mpipi_epsilon_values.append(mpipi_eps)
        
        # Calculate CALVADOS epsilon (self-interaction)
        calvados_eps = cf.epsilon(sequence, sequence)
        calvados_epsilon_values.append(calvados_eps)
        
    except Exception as e:
        print(f"  Error processing {identifier}: {e}")
        mpipi_epsilon_values.append(np.nan)
        calvados_epsilon_values.append(np.nan)

# Add the epsilon values as new columns
epsilons['mpipi_epsilon'] = mpipi_epsilon_values
epsilons['calvados_epsilon'] = calvados_epsilon_values
```

```python
epsilons.to_csv('epsilons.csv', index = False)
```
